```python
import os, time, json
from pathlib import Path

env_path = Path.cwd() / ".env"
if not env_path.is_file():
    env_path = Path.cwd() / "genai-pce" / "M1_GenAI_LLM_Foundations" / ".env"
if not env_path.is_file():
    raise FileNotFoundError(f"Could not find .env at {env_path}")

for line in env_path.read_text().splitlines():
    line = line.strip()
    if not line or line.startswith("#") or "=" not in line:
        continue
    key, value = line.split("=", 1)
    os.environ[key.strip()] = value.strip().strip("'\"")

BASE_URL = os.getenv("BASE_URL")
API_KEY = os.getenv("OPENROUTER_API_KEY")
MODEL = os.getenv("MODEL")

required = {"BASE_URL": BASE_URL, "OPENROUTER_API_KEY": API_KEY, "MODEL": MODEL}
missing = [name for name, value in required.items() if not value]
if missing:
    raise ValueError(f"Missing values in {env_path}: {', '.join(missing)}")
```

```python
from openai import OpenAI
# Create a Client
client = OpenAI(
    api_key=API_KEY, 
    base_url=BASE_URL)
```

```python
response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {
            "role": "user",
            "content": "Hello! Can you tell me a joke?"
        }
    ]
)
print(response.choices[0].message.content)
print("Finish: ",response.choices[0].finish_reason)
print("Token Usage: ",response.usage.prompt_tokens + response.usage.completion_tokens + response.usage.total_tokens)
print("Model: ", response.model)
```
